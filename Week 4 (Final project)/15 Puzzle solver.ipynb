```python
#Defining states
from itertools import permutations

row1_states = []
row2_states = []
row3_states = []
goal_row1 = tuple([_ for _ in range(1, 5)])
goal_row2 = tuple([_ for _ in range(1,9)])
goal_row3 = tuple([_ for _ in range(1, 17)])

for i in range(16):
    for j in range(16):
        for k in range(16):
            for l in range(16):
                a = [i-j, j-k, k-l, i-k, j-l, i-l]
                if 0 not in a:
                    for m in range(16):
                        if(m!=i and m!=j and m!=k and m!=l):
                            tmp = [0]*16
                            tmp[i]=1
                            tmp[j]=2
                            tmp[k]=3
                            tmp[l]=4
                            tmp[m]=16
                            row1_states.append(tuple(tmp))
                        
for i in range(4, 16):
    for j in range(4, 16):
        for k in range(4, 16):
            for l in range(4, 16):
                a = [i-j, j-k, k-l, i-k, j-l, i-l]
                if 0 not in a:
                    for m in range(4, 16):
                        if(m!=i and m!=j and m!=k and m!=l):
                            tmp = [1,2,3,4]+[0]*12
                            tmp[i]=5
                            tmp[j]=6
                            tmp[k]=7
                            tmp[l]=8
                            tmp[m]=16
                            row2_states.append(tuple(tmp))

for i in permutations([_ for _ in range(8, 16)]):
    tmp = [_ for _ in range(1,9)]+[0]*8
    for j in range(7):
        tmp[i[j]]=j+9
    tmp[i[-1]]=16
    row3_states.append(tuple(tmp))   

#Making MDPs
mdp_row1 = {_:{} for _ in row1_states}
mdp_row2 = {_:{} for _ in row2_states}
mdp_row3 = {_:{} for _ in row3_states}

#Row 1 MDP Filling
for state in mdp_row1:
    actions = ['u', 'd', 'l', 'r']
    trans = {}
    st_list = list(state)
    st_goal = tuple(state[0:4])
    for act in actions:
        next_s = list(state).copy()
        hole = next_s.index(16)
        if st_goal == goal_row1:
            trans[act]=[(1, state, 0, False)]
            continue
        if act=='l':
            if hole%4 == 0:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole-1] = next_s[hole-1], next_s[hole]
        elif act=='r':
            if hole%4 == 3:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole+1] = next_s[hole+1], next_s[hole]
        elif act=='u':
            if hole//4 == 0:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole-4] = next_s[hole-4], next_s[hole]
        elif act=='d':
            if hole//4 == 3:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole+4] = next_s[hole+4], next_s[hole]
        corr = sum([1 for i in range(4) if next_s[i]==goal_row1[i]])
        rew = -2 + 0.5*corr
        if next_s[0:4]==list(goal_row1):
            rew = 67
        trans[act]=[(1, tuple(next_s), rew, rew==1)]
    mdp_row1[state]=trans

#Row 2 MDP Filling
for state in mdp_row2:
    actions = ['u', 'd', 'l', 'r']
    trans = {}
    st_list = list(state)
    st_goal = tuple(state[0:8])
    for act in actions:
        next_s = list(state).copy()
        hole = next_s.index(16)
        if st_goal == goal_row2:
            trans[act]=[(1, state, 0, False)]
            continue
        if act=='l':
            if hole%4 == 0:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole-1] = next_s[hole-1], next_s[hole]
        elif act=='r':
            if hole%4 == 3:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole+1] = next_s[hole+1], next_s[hole]
        elif act=='u':
            if hole//4 <= 1:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole-4] = next_s[hole-4], next_s[hole]
        elif act=='d':
            if hole//4 == 3:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole+4] = next_s[hole+4], next_s[hole]
        corr = sum([1 for i in range(8) if next_s[i]==goal_row2[i]])
        rew = -5 + 0.5*corr
        if next_s[0:8]==list(goal_row2):
            rew = 140
        trans[act]=[(1, tuple(next_s), rew, rew==1)]
    mdp_row2[state]=trans

#Row 3 MDP Filling
for state in mdp_row3:
    actions = ['u', 'd', 'l', 'r']
    trans = {}
    st_list = list(state)
    st_goal = tuple(state[0:16])
    for act in actions:
        next_s = list(state).copy()
        hole = next_s.index(16)
        if st_goal == goal_row3:
            trans[act]=[(1, state, 0, False)]
            continue
        if act=='l':
            if hole%4 == 0:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole-1] = next_s[hole-1], next_s[hole]
        elif act=='r':
            if hole%4 == 3:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole+1] = next_s[hole+1], next_s[hole]
        elif act=='u':
            if hole//4 <= 2:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole-4] = next_s[hole-4], next_s[hole]
        elif act=='d':
            if hole//4 == 3:
                trans[act]=[(1, tuple(next_s), -67, False)]
                continue
            next_s[hole], next_s[hole+4] = next_s[hole+4], next_s[hole]
        corr = sum([1 for i in range(16) if next_s[i]==goal_row3[i]])
        rew = -8 + 0.5*corr
        if next_s==list(goal_row3):
            rew = 250
        trans[act]=[(1, tuple(next_s), rew, rew==1)]
    mdp_row3[state]=trans

#Value Iteration
def value_iteration(states, goal, mdp, num, gamma):
    actions=['u', 'd', 'r', 'l']
    v = {s:-10.0 for s in states}
    new_v = {s:0.0 for s in states}
    old_policy = {s:'u' for s in states}
    new_policy = {s:'d' for s in states}
    for s in states:
        if goal == s[:num]:
            v[s] = 1
            new_v[s] = 1
    tol = 1e-2

    while True:
        old_policy = new_policy.copy()
        v = new_v.copy()
        delta = 0
        for s in states:
            if list(s)[:num]==goal:
                continue
            best_val = -float('inf')
            best_act = None
            for act in actions:
                a1 = mdp[s][act][0]
                val = a1[0]*(a1[2] + gamma*v[a1[1]])
                if val > best_val:
                    best_val = val
                    best_act = act
            new_v[s] = best_val
            delta = max(delta, abs(new_v[s]-v[s]))
        if delta < tol:
            for s in states:
                if list(s)[:num]==goal:
                    continue
                best_val = -float('inf')
                best_act = None
                for act in actions:
                    a1 = mdp[s][act][0]
                    val = a1[0]*(a1[2] + gamma*new_v[a1[1]])
                    if val>best_val:
                        best_val = val
                        best_act = act
                new_policy[s] = best_act
            break
    return new_policy

#Compute policies
policy_row1 = value_iteration(row1_states, goal_row1, mdp_row1, 4, 0.9)
policy_row2 = value_iteration(row2_states, goal_row2, mdp_row2, 8, 0.9)
policy_row3 = value_iteration(row3_states, goal_row3, mdp_row3, 16, 0.9)

#Helper functions
def next_state(state, act):
    ns = list(state).copy()
    hole = ns.index(16)
    if act=='l':
        if hole%4==0: return state
        ns[hole], ns[hole-1] = ns[hole-1], ns[hole]
    elif act=='r':
        if hole%4==3: return state
        ns[hole], ns[hole+1] = ns[hole+1], ns[hole]
    elif act=='u':
        if hole//4==0: return state
        ns[hole], ns[hole-4] = ns[hole-4], ns[hole]
    elif act=='d':
        if hole//4==3: return state
        ns[hole], ns[hole+4] = ns[hole+4], ns[hole]
    return tuple(ns)

def convert_state(state, num):
    st = list(state).copy()
    for i in range(16):
        if st[i]>num and st[i]!=16:
            st[i]=0
    return tuple(st)

def print_state(state):
    for i in range(16):
        print(state[i], end=' ')
        if i in [3,7,11,15]:
            print()

#Random start example
import random
given_state = list(range(1,17))
random.shuffle(given_state)
given_state = tuple(given_state)

move_count = 0
curr_state = tuple(given_state)

while curr_state[:4]!=goal_row1:
    act = policy_row1[convert_state(curr_state,4)]
    move_count += 1
    print(act)
    curr_state = next_state(curr_state, act)
    print_state(curr_state)
print("ROW 1 DONE")

while curr_state[:8]!=goal_row2:
    act = policy_row2[convert_state(curr_state,8)]
    move_count += 1
    print(act)
    curr_state = next_state(curr_state, act)
    print_state(curr_state)
print("ROW 2 DONE")

while curr_state!=goal_row3:
    act = policy_row3[convert_state(curr_state,16)]
    move_count += 1
    print(act)
    curr_state = next_state(curr_state, act)
    print_state(curr_state)
print("PUZZLE DONE IN", move_count, "MOVES")


```
