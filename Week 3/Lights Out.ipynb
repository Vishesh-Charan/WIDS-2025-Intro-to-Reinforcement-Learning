```python
# Lights Out MDP 
import numpy as np
from collections import deque

N = 4
NUM_BITS = N * N
NUM_STATES = 1 << NUM_BITS
ACTIONS = list(range(NUM_BITS))
GAMMA = 1.0
def toggle(x, i):
    return x ^ (1 << i)

def transition(state, switch):
    a = toggle(state, 15 - switch)
    if switch > 3:
        a = toggle(a, 19 - switch)
    if switch % 4 > 0:
        a = toggle(a, 16 - switch)
    if switch % 4 < 3:
        a = toggle(a, 14 - switch)
    if switch < 12:
        a = toggle(a, 11 - switch)
    return a
# Build MDP: lo_map[state][action] = [(prob, next_state, reward)]
lo_map = {}

for state in range(NUM_STATES):
    actions = {}
    for a in ACTIONS:
        ns = transition(state, a)
        reward = 1 if ns == 0 else 0
        actions[a] = [(1.0, ns, reward)]
    lo_map[state] = actions

```

```python
# Value Iteration (Seems the Optimal One here)
V = [-1] * NUM_STATES
policy = [0] * NUM_STATES

V[0] = 0
queue = deque([0])

while queue:
    s = queue.popleft()
    for a in ACTIONS:
        ns = transition(s, a)
        if V[ns] == -1:
            V[ns] = V[s] + 1
            policy[ns] = a
            queue.append(ns)
```

```python
# Finding min moves of gurantee Solve any starting board
worst_case_moves = max(V)
print("Minimum number of moves to guarantee solving any board:",
      worst_case_moves)
```
